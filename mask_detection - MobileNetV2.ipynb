```python
import os
import numpy as np
import sklearn
```

```python
import cv2
```

```python
from keras.preprocessing import image
```

```python
categories=['with_mask','without_mask']
```

```python
data = []
for category in categories:
    path = os.path.join('train',category)
    label = categories.index(category)
    
    for file in os.listdir(path):
        img_path = os.path.join(path,file)
        img = cv2.imread(img_path)
        
        img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
        
        img = cv2.resize(img,(224,224))
        data.append([img,label])
```

```python
len(data)
```

```python
import random
```

```python
random.shuffle(data)
```

```python
X=[]
y=[]

for features,label in data:
    X.append(features)
    y.append(label)
```

```python
X=np.array(X)
y=np.array(y)
```

```python
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
X = preprocess_input(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
X_train.shape
```

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,GlobalAveragePooling2D
```

```python
base_model=MobileNetV2(weights='imagenet',include_top=False,input_shape=(224,224,3))
base_model.trainable=False

model=Sequential()
model.add(base_model)
model.add(GlobalAveragePooling2D())
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```

```python
model.save('new_mask_detection_MobileNetV2.h5')
```

```python
import numpy as np
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

def detect_face_mask(img):
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    img_reshaped = img_rgb.reshape(1, 224, 224, 3)
    
    img_preprocessed = preprocess_input(np.float32(img_reshaped))
   
    predictions = model.predict(img_preprocessed,verbose=0)
    y_pred = (predictions > 0.5).astype("int32")
    
    return y_pred[0][0]
```

```python
def draw_label(img,text,pos,bgcolor):
    text_size=cv2.getTextSize(text,cv2.FONT_HERSHEY_SIMPLEX,1,cv2.FILLED)

    end_x=pos[0] + text_size[0][0] + 2
    end_y=pos[1] + text_size[0][1] - 2

    cv2.rectangle(img,pos,(end_x,end_y),bgcolor,cv2.FILLED)
    cv2.putText(img,text,pos,cv2.FONT_HERSHEY_SIMPLEX,1,(0,0,0),1,cv2.LINE_AA)
```

```python
import cv2
haar = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
def detect_face(img):
    coods=haar.detectMultiScale(img)
    return coods
```
