```python
import os
import numpy as np
import sklearn
```

```python
import cv2
```

```python
from keras.preprocessing import image
```

```python
categories=['with_mask','without_mask']
```

```python
data=[]
for category in categories:
    path=os.path.join('train',category)

    label=categories.index(category)

    for file in os.listdir(path):
        img_path=os.path.join(path,file)
        img=cv2.imread(img_path)
        img=cv2.resize(img,(224,224))

        data.append([img,label])

    
```

```python
len(data)
```

```python
import random
```

```python
random.shuffle(data)
```

```python
X=[]
y=[]

for features,label in data:
    X.append(features)
    y.append(label)
```

```python
X=np.array(X)
y=np.array(y)
```

```python
X=X/255
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
X_train.shape
```

```python
from keras.applications.vgg16 import VGG16
```

```python
vgg=VGG16()
```

```python
vgg.summary()
```

```python
from keras import Sequential
```

```python
model=Sequential()
```

```python
for layer in vgg.layers[:-1]:
    model.add(layer)
```

```python
model.summary()
```

```python
for layer in model.layers:
    layer.trainable=False
```

```python
model.summary()
```

```python
from keras.layers import Dense
```

```python
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```

```python
model.save('new_VGG16.h5')
```

```python
cap=cv2.VideoCapture(0)
```

```python
while True:
    ret,frame=cap.read()

    resized_frame = cv2.resize(frame, (224, 224))
    y_pred = detect_face_mask(resized_frame)

    coods=detect_face(cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY))
    
    for x,y,w,h in coods:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),2)

    if y_pred==0:
        draw_label(frame,'Mask',(30,30),(0,255,0))

    else:
        draw_label(frame,'No Mask',(30,30),(0,0,255))

    cv2.imshow('window',frame)
    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

cap.release() 
cv2.destroyAllWindows()
```

```python
def detect_face_mask(img):
    predictions = model.predict(img.reshape(1, 224, 224, 3), verbose=0)
    y_pred = (predictions > 0.5).astype("int32")
    
    return y_pred[0][0]
```

```python
def draw_label(img,text,pos,bgcolor):
    text_size=cv2.getTextSize(text,cv2.FONT_HERSHEY_SIMPLEX,1,cv2.FILLED)

    end_x=pos[0] + text_size[0][0] + 2
    end_y=pos[1] + text_size[0][1] - 2

    cv2.rectangle(img,pos,(end_x,end_y),bgcolor,cv2.FILLED)
    cv2.putText(img,text,pos,cv2.FONT_HERSHEY_SIMPLEX,1,(0,0,0),1,cv2.LINE_AA)
```

```python
import cv2
haar = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
def detect_face(img):
    coods=haar.detectMultiScale(img)
    return coods
```
